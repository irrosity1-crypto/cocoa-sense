Exploration only: all logic lives in `src/cocoasense`. Run `uv run cocoasense validate` first, then open with `uv run --extra notebook jupyter lab`.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from cocoasense.config import load_config
from cocoasense.pipeline import load_monthly

cfg = load_config()
monthly = load_monthly(cfg)
monthly.describe()
```

```python
price = monthly['price_usd_t']
ax = price.plot(logy=True, figsize=(10, 4), color='black')
ax.axvspan('2016-01', '2019-12', color='orange', alpha=0.3, label='validation')
ax.axvspan('2020-01', price.index.max(), color='steelblue', alpha=0.3, label='test')
ax.set_ylabel('US$ per tonne (log)'); ax.legend(); plt.show()
```

```python
r = np.log(price).diff().dropna()
fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
r.plot(ax=axes[0], color='grey'); axes[0].set_ylabel('monthly log return')
r.rolling(12).std().plot(ax=axes[1], color='firebrick'); axes[1].set_ylabel('12-month volatility')
plt.show()
pd.DataFrame({'sd_monthly_log_return': [r[:'2019'].std(), r['2020':'2022'].std(), r['2023':].std()]},
             index=['to 2019', '2020-2022', '2023 on'])
```

```python
ax = monthly['ghs_per_usd'].dropna().plot(logy=True, figsize=(10, 4))
ax.set_ylabel('cedis per US$ (log)'); plt.show()
monthly.loc['2007-04':'2007-10', 'ghs_per_usd']  # no redenomination break
```

```python
from statsmodels.graphics.tsaplots import plot_acf
plot_acf(r[:'2019'], lags=24); plt.title('ACF of monthly log returns, training era'); plt.show()
```
